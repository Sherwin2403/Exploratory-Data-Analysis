```python
df = df.drop(columns = ['Id'])
```

```python
num_cols = df.select_dtypes(include = ['int64', 'float64']).columns
num_cols
```

```python
cat_cols = df.select_dtypes(include = ['object', 'category']).columns
cat_cols
```

```python
for col in num_cols:
  plt.figure( figsize = (6, 3))
  sns.boxplot(df[col])
  plt.title(col)
  plt.show()
```

```python
print(df[num_cols].skew())
```

```python
# Distribution plots for all numerical columns

for col in num_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
# Count plot for wine quality

plt.figure(figsize=(7, 4))
sns.countplot(x='quality', data=df)
plt.title('Distribution of Wine Quality')
plt.xlabel('Quality')
plt.ylabel('Count')
plt.show()
```

```python
# Mean, Median and Mode of numerical columns

for col in num_cols:
    print("\nColumn:", col)
    print("Mean   :", df[col].mean())
    print("Median :", df[col].median())
    print("Mode   :", df[col].mode()[0])
```

```python
# Measures of dispersion

dispersion = pd.DataFrame({
    'Minimum': df[num_cols].min(),
    'Maximum': df[num_cols].max(),
    'Range': df[num_cols].max() - df[num_cols].min(),
    'Variance': df[num_cols].var(),
    'Standard Deviation': df[num_cols].std()
})

dispersion
```

```python
# Box plot of wine quality

plt.figure(figsize=(7, 4))
sns.boxplot(x='quality', y='alcohol', data=df)
plt.title('Alcohol Content by Wine Quality')
plt.xlabel('Quality')
plt.ylabel('Alcohol')
plt.show()
```

```python
# Scatter plot for Alcohol vs Quality

plt.figure(figsize=(7, 4))
plt.scatter(df['alcohol'], df['quality'])
plt.title('Scatter Plot: Alcohol vs Quality')
plt.xlabel('Alcohol')
plt.ylabel('Quality')
plt.show()
```

```python
# Scatter plot for Sulphates vs Quality

plt.figure(figsize=(7, 4))
plt.scatter(df['sulphates'], df['quality'])
plt.title('Scatter Plot: Sulphates vs Quality')
plt.xlabel('Sulphates')
plt.ylabel('Quality')
plt.show()
```

```python
# Box plots to compare numerical variables across wine quality

selected_cols = ['alcohol', 'volatile acidity', 'citric acid',
                 'sulphates', 'total sulfur dioxide']

for col in selected_cols:
    plt.figure(figsize=(7, 4))
    sns.boxplot(x='quality', y=col, data=df)
    plt.title(f'{col} vs Wine Quality')
    plt.xlabel('Quality')
    plt.ylabel(col)
    plt.show()
```

```python
# Pair plot for selected numerical variables

pair_cols = [
    'fixed acidity',
    'volatile acidity',
    'citric acid',
    'alcohol',
    'quality'
]

sns.pairplot(df[pair_cols], kind='reg')
plt.show()
```

```python
# Pearson correlation between Alcohol and Quality

from scipy import stats

corr = stats.pearsonr(df['alcohol'], df['quality'])

print("p-value:", corr[1])
print("Correlation coefficient:", corr[0])
```

```python
# Pearson correlation matrix

correlation = df.corr(method='pearson', numeric_only=True)

correlation
```

```python
# Visualize correlation matrix using a heatmap

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Wine Quality Dataset')
plt.show()
```

```python
# Correlation of all numerical variables with wine quality

quality_corr = correlation['quality'].sort_values(ascending=False)

print(quality_corr)
```

```python
# Plot correlations with quality

plt.figure(figsize=(8, 5))

quality_corr.drop('quality').sort_values().plot(kind='barh')

plt.title('Correlation of Features with Wine Quality')
plt.xlabel('Correlation Coefficient')
plt.ylabel('Feature')
plt.show()
```
