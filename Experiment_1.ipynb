```python
arrays = [
    ["bar", "bar", "baz", "baz", "foo", "foo", "qux", "qux"],
    ["one", "two", "one", "two", "one", "two", "one", "two"],
]
```

```python
tuples = list(zip(*arrays))
tuples
```

```python
index = pd.MultiIndex.from_tuples(tuples, name = ['first', 'second'])
index
```

```python
s = pd.Series(np.random.randn(8), index = index)
s
```

```python
#When you want every pairing of the elements in two iterables, it can be easier to use the MultiIndex.from_product() method:

iterables = [['bar', 'baz', 'foo', 'qux'], ['one', 'two']]

pd.MultiIndex.from_product(iterables, names = ['first', 'second'])
```

```python
#You can also construct a MultiIndex from a DataFrame directly, using the method MultiIndex.from_frame(). This is a complementary method to MultiIndex.to_frame().

df = pd.DataFrame(
    [['bar', 'one'], ['bar', 'two'], ['foo', 'one'], ['foo', 'two']],
    columns = ['first', 'second']
)

pd.MultiIndex.from_frame(df)
```

```python
#The MultiIndex keeps all the defined levels of an index, 
# even if they are not actually used. When slicing an index, you may notice this. For example:

df.columns.levels

df[['foo', 'qux']].columns.levels
```

```python
#This is done to avoid a recomputation of the levels in order to make slicing highly performant. 
# If you want to see only the used levels, you can use the get_level_values() method.
df[['foo', 'qux']].columns.to_numpy()
```

```python
df[['foo', 'qux']].columns.get_level_values(0)
```

```python
#To reconstruct the MultiIndex with only the used levels, 
# the remove_unused_levels() method may be used.

new_mi = df[['foo', 'qux']].columns.remove_unused_levels()

new_mi.levels
```

```python
df1 = pd.DataFrame(
    {
        'A' : ['A0', 'A1', 'A2', 'A3'], 
        'B' : ['B0', 'B1', 'B2', 'B3'],
        'C' : ['C0', 'B1', 'C2', 'C3'],
        'D' : ['D0', 'D1', 'D2', 'D3']
    }, index = [0, 1, 2, 3]
)

df2 = pd.DataFrame(
    {
        "A": ["A4", "A5", "A6", "A7"],
        "B": ["B4", "B5", "B6", "B7"],
        "C": ["C4", "C5", "C6", "C7"],
        "D": ["D4", "D5", "D6", "D7"],
    },
    index=[4, 5, 6, 7],
)

df3 = pd.DataFrame(
    {
        "A": ["A8", "A9", "A10", "A11"],
        "B": ["B8", "B9", "B10", "B11"],
        "C": ["C8", "C9", "C10", "C11"],
        "D": ["D8", "D9", "D10", "D11"],
    },
    index=[8, 9, 10, 11],
)

frames = [df1, df2, df3]

result = pd.concat(frames)

result
```

```python
df4 = pd.DataFrame(
    {
        "B": ["B2", "B3", "B6", "B7"],
        "D": ["D2", "D3", "D6", "D7"],
        "F": ["F2", "F3", "F6", "F7"],
    },
    index=[2, 3, 6, 7],
)

result = pd.concat([df1, df4], axis=1)

result
```

```python
#join='inner' takes the intersection of the axis values.

result = pd.concat([df1, df4], axis = 1, join = 'inner')
result
```

```python
#To perform an effective “left” join using the exact index 
# from the original DataFrame, result can be reindexed.

result = pd.concat([df1, df4], axis = 1).reindex(df1.index)

result
```
