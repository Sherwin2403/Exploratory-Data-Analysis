```python
import pandas as pd
```

```python
continents = ['North America', 'North America', 'Europe', 'Europe']
countries = ['USA', 'Canada', 'Germany', 'France']
cities = ['New York', 'Toronto', 'Berlin', 'Paris']
sales = [500000, 320000, 410000, 290000]

index = pd.MultiIndex.from_arrays(
    [continents, countries, cities], names = ['Continent', 'Country', 'City']
)

df = pd.DataFrame({'Sales_USD': sales}, index = index)
print(df)
```

```python
north_america_sales = df.loc['North America']
print(north_america_sales)
```

```python
toronto_sales = df.loc['North America', 'Canada', 'Toronto',]
print(toronto_sales)
```

```python
continents_totals = df.groupby(level = "Continent").sum()
print(continents_totals)
```

```python
df_inventory = pd.DataFrame({
    'Product_ID': [101, 102, 103],
    'Qty': [10, 20, 30]
})
```

```python
df_sales = pd.DataFrame({
    'Product_ID' : [101, 102, 104],
    'Qty': [5, 1, 2]
})
```

```python
inner_df = pd.merge(df_inventory, df_sales, on = 'Product_ID', how ='inner')
print(inner_df)
```

```python
outer_df = pd.merge(df_inventory, df_sales, on = 'Product_ID', how = 'outer')
print(outer_df)
```

```python
df_ny = pd.DataFrame({'Item': ['Apple', 'Banana'], 'Sales' : [150, 90]})

df_lon = pd.DataFrame({'Item': ['Orange', 'Banana'], 'Sales' : [200, 110]})

df_global = pd.concat([df_ny, df_lon], keys = ['New York', 'London'])
print(df_global)
```

```python
df_indexed = df_global.set_index('Item', append=True).sort_index()

idx = pd.IndexSlice
banana_sales = df_indexed.loc[idx[:, :, 'Banana'], :]
print(banana_sales)
```
