```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
```

```python
data = {
 'Age': [25, 30, 45, 35, 22],
 'Salary': [50000, 64000, 120000, 85000, 1200000] # 1,200,000 is an outlier
}
```

```python
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")
```

```python
min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
 min_max_scaler.fit_transform(df),
 columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")
```

```python
standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
 standard_scaler.fit_transform(df),
 columns=df.columns
)
print("Standardized Data:\n", df_standard, "\n")
```

```python
# Drop the ID column permanently from your dataframe
df = df.drop(columns=['student_id'])
```

```python
# 1. Automatically separate columns by data type
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

# 2. Print the results to verify
print("Numerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
# Show value counts including the missing (NaN) count
print(df['parental_education'].value_counts(dropna=False))
```

```python
# Creates a 1/0 flag column
df['is_missing'] = df['parental_education'].isna().astype(int)
```

```python
df['parental_education'] = df['parental_education'].fillna('Missing')
```

```python
education_map = {
    'Missing': 0,
    'High School': 1,
    'Bachelors': 2,
    'Masters': 3,
    'PhD': 4
}
```

```python
df['parental_education_encoded'] = df['parental_education'].map(education_map)
```

```python
print(df['parental_education_encoded'].skew())
```

```python
df['parental_education_encoded'].hist(bins=20)
```

```python
# Fill missing values with the median
median_val = df['parental_education_encoded'].median()
df['parental_education_encoded'] = df['parental_education_encoded'].fillna(median_val)
```
