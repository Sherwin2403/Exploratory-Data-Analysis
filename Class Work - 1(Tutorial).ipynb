```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
data = {
    'EmployeeID': range(101, 601),
    'Name': [f'Employee_{i}' for i in range(1, 501)],
    'Department': np.random.choice(['HR', 'IT', 'Finance', 'Marketing', 'Operations'], size=500),
    'Age': np.random.randint(22, 60, size=500),
    'Experience': np.random.randint(1, 35, size=500),
    'Salary': np.random.randint(40000, 150000, size=500),
    'PerformanceRating': np.random.choice([1, 2, 3, 4, 5], size=500),
    'RemoteWork': np.random.choice(['Yes', 'No'], size=500)
}
```

```python
df_synthetic = pd.DataFrame(data)
```

```python
duplicates = df_synthetic.sample(n=12, random_state=42)
df_synthetic = pd.concat([df_synthetic, duplicates], ignore_index=True)
```

```python
df_synthetic = df_synthetic.iloc[:500].copy()
```

```python
df_synthetic.loc[df_synthetic.sample(n=15, random_state=1).index, 'Age'] = np.nan
df_synthetic.loc[df_synthetic.sample(n=25, random_state=2).index, 'Salary'] = np.nan
```

```python
df_synthetic.to_csv('employee_salary.csv', index=False)
```
